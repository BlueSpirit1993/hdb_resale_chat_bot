```python
import pickle
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from lightgbm import LGBMRegressor

# Define your column groups
numerical_cols = ["floor_area_sqm", "GDP", "remaining_lease_year", "walk_distance_to_school", "walk_distance_to_mrt"]
date_cols = ["month_sin", "month_cos", "salesYear_fr_2017"]

# Create preprocessor
preprocessor = ColumnTransformer(transformers=[
    ('scaled_num', Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', RobustScaler())
    ]), numerical_cols),

    ('keep_scaled', 'passthrough', numerical_cols),
    ('keep_dates', 'passthrough', date_cols)
])

# Use the best parameters you found
best_lgbm = LGBMRegressor(
    learning_rate=0.1,
    max_depth=15,
    n_estimators=300,
    num_leaves=100,
    objective='regression',
    random_state=42
)

# Create the pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', best_lgbm)
])

# Fit the model
pipeline.fit(X, y)

# Save the model
with open("best_lgbm_pipeline.pkl", "wb") as f:
    pickle.dump(pipeline, f)

# Or use joblib (more efficient for larger models)
# import joblib
# joblib.dump(pipeline, "best_lgbm_pipeline.pkl")
```
